# Histórico CSV e janela de conversa

Base: Aula 01 p. 40 e Aula 02 pp. 14–15. Exercício autoral. Conte mensagens e turnos separadamente: dois turnos + system = cinco mensagens; quatro turnos + system = nove. Compare a remoção literal de vírgulas pedida no slide com armazenamento CSV que preserva o texto.

```python
from pathlib import Path
import sys,tempfile
RAIZ=Path.cwd()
if RAIZ.name=="notebook": RAIZ=RAIZ.parent
sys.path.insert(0,str(RAIZ))
from src.historico import janela,salvar_csv,carregar_csv
mensagens=[("system","Seja claro")]
for i in range(5):
    mensagens += [("human",f"Pergunta {i}"),("assistant",f"Resposta {i}")]
assert len(janela(mensagens,2))==5
assert len(janela(mensagens,4))==9
janela(mensagens,2)
```

```python
original=[("system","Contexto"),("human",'Bom, mas "caro".\nChegou hoje'),("assistant","Entendi")]
with tempfile.TemporaryDirectory() as td:
    caminho=Path(td)/"historico.csv"
    salvar_csv(caminho,original)
    assert carregar_csv(caminho)==original
    salvar_csv(caminho,original,remover_virgulas=True)
    assert carregar_csv(caminho)[1][1]==original[1][1].replace(",", "")
```
