# Templates placeholder e esquema validado

Base: Aula 02 pp. 17–36. Formatar um prompt não chama o modelo. Pedir JSON não valida tipos ou valores. Estes exemplos testam formatação e Pydantic offline.

```python
from langchain_core.prompts import PromptTemplate,ChatPromptTemplate
texto=PromptTemplate.from_template("Classifique {comentario} sobre {produto}")
assert "notebook" in texto.format(comentario="Muito bom",produto="notebook")
chat=ChatPromptTemplate.from_messages([
 ("system","Classifique o sentimento sobre {produto}"),
 ("placeholder","{historico}"),
 ("human","{comentario}")])
formatado=chat.invoke({"produto":"notebook","historico":[],"comentario":"Muito bom"})
assert len(formatado.to_messages())==2
formatado.to_messages()
```

```python
from typing import Literal
from pydantic import BaseModel,Field,ValidationError
class Resposta(BaseModel):
    sentimento: Literal["positivo","negativo"]
    confianca: float=Field(ge=0,le=1)
certa=Resposta.model_validate_json('{"sentimento":"positivo","confianca":0.8}')
assert certa.sentimento=="positivo"
try:
    Resposta.model_validate({"sentimento":"neutro","confianca":1.4})
except ValidationError as erro:
    assert len(erro.errors())==2
    print("Dois erros identificados")
```
